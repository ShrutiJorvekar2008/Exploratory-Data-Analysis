```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cardiotocographic.csv")
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(), inplace=True)
```

```python
plt.figure(figsize=(15,7))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
df.describe()
```

```python
df.quantile([0.25, 0.75])
```

```python
df.hist(figsize=(15,10), bins=20)
plt.suptitle("Distribution of Numerical Variables")
plt.show()
```

```python
plt.figure(figsize=(15,7))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("Boxplot of Dataset Variables")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df.sample(200))
plt.show()
```

```python
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.title("Violin Plot of Numerical Features")
plt.show()
```

##### Strong correlations among certain fetal heart rate variables suggest they may jointly influence fetal health assessment.

Skewed distributions indicate the presence of abnormal fetal conditions in a subset of observations.
